# Lektion 2: Der Computer trifft Entscheidungen

Bisher laufen unsere Programme einfach von oben nach unten durch. Aber intelligente Programme müssen Entscheidungen treffen:
* „Ist das Passwort richtig?“ → Zugriff erlauben.
* „Ist das Passwort falsch?“ → Zugriff verweigern.

Dazu nutzen wir `if` (wenn) und `else` (sonst).

```python
# Wir simulieren eine Alterskontrolle im Kino
alter = int(input("Wie alt bist du? "))

if alter >= 16:
    print("Du darfst den Horrorfilm sehen.")
    print("Viel Spaß!")
else:
    print("Du bist leider noch zu jung.")
    print("Hier ist ein Ticket für den Disney-Film.")

print("Der nächste Kunde bitte.") # Diese Zeile wird IMMER ausgeführt (keine Einrückung)
```

## Achtung: Die Einrückung (Indentation)

In Python ist es extrem wichtig, wie weit der Code eingerückt ist.
Alles, was **eingerückt** ist, gehört zur Entscheidung dazu.
Alles, was wieder vorne an der Zeile steht, gehört nicht mehr dazu.

**Vergleichsoperatoren:**
* `==`  (Ist gleich? Wichtig: Doppeltes Gleichzeichen!)
* `!=`  (Ist ungleich?)
* `<`   (Kleiner als)
* `>`   (Größer als)
