# Lektion 2: Der Computer trifft Entscheidungen

Bisher laufen unsere Programme einfach von oben nach unten durch. Aber intelligente Programme müssen Entscheidungen treffen:
* „Ist das Passwort richtig?“ → Zugriff erlauben.
* „Ist das Passwort falsch?“ → Zugriff verweigern.

Dazu nutzen wir `if` (wenn) und `else` (sonst).

```python
# Wir simulieren eine Alterskontrolle im Kino
alter = int(input("Wie alt bist du? "))

if alter >= 16:
    print("Du darfst den Horrorfilm sehen.")
    print("Viel Spaß!")
else:
    print("Du bist leider noch zu jung.")
    print("Hier ist ein Ticket für den Disney-Film.")

print("Der nächste Kunde bitte.") # Diese Zeile wird IMMER ausgeführt (keine Einrückung)
```

## Achtung: Die Einrückung (Indentation)

In Python ist es extrem wichtig, wie weit der Code eingerückt ist.
Alles, was **eingerückt** ist, gehört zur Entscheidung dazu.
Alles, was wieder vorne an der Zeile steht, gehört nicht mehr dazu.

**Vergleichsoperatoren:**
* `==`  (Ist gleich? Wichtig: Doppeltes Gleichzeichen!)
* `!=`  (Ist ungleich?)
* `<`   (Kleiner als)
* `>`   (Größer als)

## 🎮 Bonus-Level: Dein eigenes Text-Adventure

Du hast jetzt gelernt, wie der Computer Entscheidungen trifft. Jetzt werden wir kreativ!
Kennst du Spiele oder Bücher, bei denen du entscheiden musst, wie die Geschichte weitergeht? ("Gehst du durch die linke oder die rechte Tür?")

Wir programmieren jetzt ein kleines **Text-Adventure**.

**Die Herausforderung:**
Wir müssen `if`-Abfragen **verschachteln**. Das bedeutet: Eine Entscheidung findet *innerhalb* einer anderen Entscheidung statt.

```python
# Ein kleines Beispiel zum Ausprobieren:

print("Du stehst vor einer dunklen Höhle.")
entscheidung = input("Möchtest du hineingehen? (ja/nein): ")

if entscheidung == "ja":
    print("Du gehst mutig hinein. Es tropft von der Decke.")
    
    # ACHTUNG: Hier beginnt die zweite Entscheidung (eingerückt!)
    licht = input("Hast du eine Taschenlampe dabei? (ja/nein): ")
    
    if licht == "ja":
        print("Du schaltest sie an und findest eine Schatztruhe! 🏆")
    else:
        print("Es ist zu dunkel. Du stößt dir den Zeh und läufst weg. 🤕")

elif entscheidung == "nein":
    print("Du bleibst lieber draußen in der Sonne. Sicher ist sicher. ☀️")

else:
    print("Das habe ich nicht verstanden. Bitte antworte mit 'ja' oder 'nein'.")
```

### ⚠️ Wichtig: Die „Treppe“ (Einrückung)

Schau dir den Code oben genau an.
1. Das zweite `if` (Taschenlampe) ist **doppelt eingerückt**.
2. Es gehört nämlich zum ersten `if` (in die Höhle gehen) dazu.
3. Nur wer in die Höhle geht, wird überhaupt nach der Taschenlampe gefragt.

Das nennt man **Verschachtelung**. Achte genau auf die Leerzeichen am Zeilenanfang!
