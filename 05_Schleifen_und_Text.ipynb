# Lektion 3: Wiederholungen (Loops)

Stell dir vor, du sollst in einem Buch jeden Buchstaben 'e' zählen. Du gehst Buchstabe für Buchstabe durch.
Genau das muss der Computer beim Cäsar-Verfahren auch tun: Jeden Buchstaben einzeln anschauen und verschieben.

Dafür gibt es die `for`-Schleife.

```python
mein_text = "Hallo"

print("Wir buchstabieren:")

# Die Variable 'buchstabe' nimmt nacheinander jeden Wert aus 'mein_text' an.
for buchstabe in mein_text:
    print(buchstabe)
    print("---") # Nur zur Deko

print("Fertig!")
```

**Was passiert hier?**
Der Computer nimmt das erste Zeichen ('H'), speichert es in `buchstabe` und führt den Codeblock aus.
Dann nimmt er das nächste Zeichen ('a'), und so weiter, bis der Text zu Ende ist.

```python
# Wir suchen nach dem Buchstaben 'a'
text = "Bananen sind lecker"
zaehler = 0

for zeichen in text:
    if zeichen == "a":
        print("Ein a gefunden!")
        zaehler = zaehler + 1

print("Anzahl der a:", zaehler)
```
