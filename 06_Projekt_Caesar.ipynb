## Das Z-Problem
Wenn wir 'z' (Stelle 25) um 1 verschieben, landen wir bei 26. Das Alphabet hat aber nur Indizes von 0 bis 25.
Wir müssen also "im Kreis" rechnen, wie bei einer Uhr. Nach 23 Uhr kommt 0 Uhr.
In Python macht das der Modulo-Operator `%`. Er gibt den Rest beim Teilen an.

`26 % 26 = 0` -> Wir landen wieder beim 'a'.
